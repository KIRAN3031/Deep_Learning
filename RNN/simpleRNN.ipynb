# Simple RNN

## Importing the neccessary libraries

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

## Input Sentence

```python
sentence = ['I', 'Love', 'Deep', 'Learning']
print(sentence)
```

## Word Embeddings

```python
word_embeddings = {
    'I':0.1,
    'Love':0.2,
    "Deep":0.3,
    'Learning':0.4
}
print(word_embeddings)
```

## Prepare Sequence Data

```python
X = np.array([
    [
        [word_embeddings["I"]],
        [word_embeddings["Love"]],
        [word_embeddings["Deep"]]
    ]
])

print(X)
```

```python
y = np.array([word_embeddings['Learning']])

print(y)
print("Input Shape : ",X.shape)
print("Output Shape :",y.shape)
```

## Building SimpleRNN Model

```python
model = Sequential()
model.add(SimpleRNN(
    units = 5,
    activation = 'tanh',
    input_shape = (3, 1)))
model.add(Dense(1, activation = 'linear'))
```

## Compile the model

```python
model.compile(optimizer = 'adam', loss = 'mse')
```

## Training the model

```python
model.fit(X, y, epochs = 1000, verbose = 1)
```

## Prediction

```python
test_sequence = np.array([
    [
        [0.1],
        [0.2],
        [0.3]
    ]
])
prediction = model.predict(test_sequence)
print("prediction : ", prediction[0][0])
```

## convert Prediction to Closest Word

```python
predicted_number = prediction[0][0]

closest_word = min(word_embeddings, key = lambda word : abs(word_embeddings[word] - predicted_number))

print("Predicted Word : ", closest_word)
```
