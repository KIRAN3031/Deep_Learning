### Text Preprocessing in NLP
    Before training NLP models, we must clean the text.
    This process is call Text Preprocessing

#### 1. Tokenization
    Breaking text into smaller units (tokens).

    It involves dividing a Textual input into smaller units known as tokens.
    These Tokens canbe in the form of words, characters, sub-words, or sentences.
    It helps in imporing interpretability of text by different models

```python
from nltk.tokenize import word_tokenize

## Words Tokenizer

text = "I love learning Natural Language Processing"
tokens = word_tokenize(text)
print(tokens)

## Character level

char_tokens = list(text)
print(char_tokens)
```

#### 2. Lowercasing
    Converting the text into lowercase

```python
text = text.lower()
print(text)
```

#### 3. Stopword Removal

    Stopwords are common words with little meaning.

    Example:

    is, the, a, an, and, of

    Example:
    I love the NLP course

    After removing stopwords:

    love NLP course

##### Tasks that benefit from stopword removal:

    Text classification and sentiment analysis
    Information retrieval and search engines
    Topic modelling and clustering
    Keyword extraction

##### Tasks that require preserving stopwords:

    Machine translation (maintains grammatical structure)

    Text summarization (preserves sentence coherence)
    Question-answering systems (syntactic relationships matter)

    Grammar checking and parsing

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
print(stop_words)

print("After Stopword Removal")
filtered_tokens = [token for token in tokens if token not in stop_words]
print(filtered_tokens)
```

#### Stemming

    Stemming reduces words to their root form.
    Stemming is an important text-processing technique that reduces words to their base or root form by removing prefixes and suffixes.
    This process standardizes words which helps to improve the efficiency and effectiveness of various natural language processing (NLP)
    tasks.

    Examples of stemming for the word "like":
    "likes" -+ "like"

    "liked" -+ "like"

    "likely" -+ "like"

    "liking" - "like"

```python
from nltk.stem import PorterStemmer
porter_stemmer=PorterStemmer()
words=["Running","Jumps","liked","funning","playing", "Happy"]
stemmed_words=[porter_stemmer.stem(word) for word in words]
print("Original Wordss:",words)
print("Stemmed Words", stemmed_words)
```

```python
#Snowball Stemmer : is an enhanced version of the Porter Stemmer, Supports multiple languages.
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer (language='english')
words_to_stem=["running","Jumped","Happily","Quickly","Foxes"]
stemmed_words=[snowball_stemmer. stem(word) for word in words_to_stem]
print("Original Words:",words_to_stem)
print("Stemmed Words:",stemmed_words)
```

```python
#Lancaster Stemmer: is known for being more aggressive and faster than other stemmers.
from nltk.stem import LancasterStemmer
lancaster_stemmer=LancasterStemmer()
words_to_stem=["running","Jumped","Happily","Quickly", "Foxes"]
stemmed_words=[lancaster_stemmer.stem(word) for word in words_to_stem]
print("Original Words:",words_to_stem)
print("Stemmed Words:",stemmed_words)
```

#### Lemmatization

    Lemmatization produces meaningful root words.

    Example

    better -+ good

    running - run

    Lemmatization is an important text pre-processing technique in Natural Language Processing (NLP) that reduces words to their base
    form known as a "lemma." For example, the lemma of "running" is "run" and "better" becomes "good."
    Unlike stemming which simply removes prefixes or suffixes, it considers the word's meaning and part of speech (POS) and ensures
    that the base form is a valid word. This makes lemmatization more accurate as it avoids generating non-dictionary words.
    Used for:

    Improves accuracy

    Reduced Data Redundancy

    Better NLP Model Performance

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
print(lemmatizer.lemmatize("better",pos='a'))
print(lemmatizer.lemmatize("running",pos='v'))
```

#### Text Representation Techniques

    Machines cannot understand text.
    They understand numbers.

    So we convert text -+ numbers.

##### Bag of Words (BoW)

BoW represents text as word frequency vectors.


In Natural Language Processing (NLP) text data needs to be converted into numbers so that machine learning algorithms can
understand it. One common method to do this is Bag of Words (BoW) model. It turns text like sentence, paragraph or document into a
collection of words and counts how often each word appears but ignoring the order of the words. It does not consider the order of the
words or their grammar but focuses on counting how often each word appears in the text.
This makes it useful for tasks like text classification, sentiment analysis and clustering.

#### Step 1: Preprocessing the Text

Clean the text

V Convert to lowercase

Remove special characters

```python
import re
dataset = []

text = [
    "I love coding",
    "I love python$3",
    "Python is         fun"
]

for sentence in text:
  sentence = sentence.lower() # lower case
  # remove symbols, extra space
  sentence = re.sub(r'\W', ' ', sentence)
  sentence = re.sub(r'\s+', ' ', sentence)
  sentence = sentence.strip()
  dataset.append(sentence)

dataset
```

#### Step 2: Counting word frequencies
Count how many times each word appears

```python
from collections import defaultdict
word_freq = defaultdict(int)
for sentence in dataset:
  words = sentence.split()
  for word in words:
    word_freq[word] += 1
print(word_freq)
```

#### Step 3: Selecting most frequent words
Let's take top 3 words

```python
import heapq
k = 3
top_words = heapq.nlargest(k, word_freq, key=word_freq.get)
print(top_words)
```

#### Step 4: Build Bag of Words Matrix

Now we create a matrix:

Rows = sentences

Columns = frequent words

Value =1 (present) or O (absent)

```python
x = []
for sentence in dataset:
  row = []
  for word in top_words:
    if word in sentence:
      row.append(1)
    else:
      row.append(0)
  x.append(row)
x
```

#### N-grams

N-grams capture word sequences.

N-gram is a technique that splits text into sequences of N consecutive words to capture context and word order for machine learning models.

N-gram = group of N consecutive words

1 word -+ Unigram
2 words - Bigram
3 words - Triaram

#### Key Idea

N-grams capture word relationships

BoW - "love", "coding"

Bigram -+ "love coding" V (more meaningful)

#### Simple Example

Sentence:

"I love coding"

Unigrams (N = 1)

["I", "love", "coding"]


Bigrams (N = 2)

["I love", "love coding"]


Trigrams (N = 3)

["I love coding"]

```python
#Step 1: Preprocessing the Text
import re
text=[
"I love coding",
"I love Python",
"Python coding is fun"
]

dataset=[]
for sentence in text:
    sentence=sentence.lower()
    sentence=re.sub(r'\W',' ',sentence)
    sentence=re.sub(r'\s+',' ',sentence)
    dataset.append(sentence)

print(dataset)
```

```python
#Step 2: Create N-Grams
# Function to generate N -Grams
def generate_ngrams(sentence, n):
    words = sentence.split()
    ngrams=[]
    for i in range(len(words)-n+1):
        ngrams.append(" ".join(words[i:i+n]))
    return ngrams
```

```python
#Example: Bigrams(N=2)
for sentence in dataset:
    bigrams=generate_ngrams(sentence,2)
    print(bigrams)

#Example: Trigrams(N=3)
for sentence in dataset:
    trigrams=generate_ngrams(sentence,3)
    print(trigrams)
```

```python
#Step 3: Count N-Gram Frequencies
from collections import defaultdict
ngram2count=defaultdict(int)

for sentence in dataset:
    ngram = generate_ngrams(sentence, 2)
    for gram in ngram:
        ngram2count[gram]+=1

print(ngram2count)
```

```python
#Step 4: Convert into feature vector (Like Bow):
#select top N-Grams:
import heapq
freq_ngrams=heapq.nlargest(2,ngram2count,key=ngram2count.get)
print(freq_ngrams)
```

```python
# Build Matrix
X = []

for sentence in dataset:
    vector = []
    sentence_ngrams = generate_ngrams(sentence, 2)

    for gram in freq_ngrams:
        if gram in sentence_ngrams:
            vector.append(1)
        else:
            vector.append(0)

    X.append(vector)

print(X)
```
